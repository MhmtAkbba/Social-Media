Data Normailization

Data normalization is the process of organizing data in a database to reduce redundancy and improve data integrity by structuring tables and establishing relationships according to specific rules.
Here's a more detailed explanation:
Purpose:
The primary goal of data normalization is to eliminate redundancy and ensure data consistency, making it easier to manage and query the database

```
# Step 1: Clean 'Interests' column (convert string into list)
df["Interests"] = df["Interests"].apply(lambda x: [i.strip().strip("'") for i in x.split(",")])
```

```

# Step 2: Explode the 'Interests' column to get one row per interest
df = df.explode("Interests")
```

The code you've provided creates an SQLite database table named Social if it doesn't already exist.

```
# CREATING THE TABLE
import sqlite3

conn = sqlite3.connect('Social.db')
print("Opened database successfully");

conn.execute('''
CREATE TABLE IF NOT EXISTS Social(UserID INTEGER,
                      Name text,
                      Gender text,
                      DOB date, Interests Text, City text, Country text);''')

conn.commit()

print("Table created successfully");
```

The code you've written loops through a Pandas DataFrame and inserts each row into an SQLite database.

```
# Loop through the DataFrame and insert each row into the database
for index, row in df.iterrows():
    conn.execute('''
        INSERT INTO Social (
            UserID, Name, Gender, DOB, Interests, City, Country
        ) VALUES (?, ?, ?, ?, ?, ?, ?)
    ''', (
        int(row['UserID']),  # UserID (Integer)
        row['Name'],         # Name (Text)
        row['Gender'],       # Gender (Text)
        row['DOB'],          # DOB (Date)
        row['Interests'],    # Interests (Text)
        row['City'],         # City (Text)
        row['Country']       # Country (Text)
    ))

# Commit the transaction to save all changes
conn.commit()
```

This analysis examines the distribution of interests across different countries and identifies how many users have each interest in those countries.

This analysis allows us to see how different interests are distributed across various countries.

It provides valuable insights into which countries are more engaged with specific interests, helping to inform marketing strategies or international studies.

By sorting by User_Count, we can identify popular interests that have a high number of users in specific countries.


```
pd.read_sql_query('''SELECT  Interests, Country, COUNT(*) AS User_Count
FROM Social
GROUP BY Interests, Country
ORDER BY Interests, User_Count DESC;''', conn)
```

This analysis identifies the number of users with an interest in Science across different countries.

This analysis helps us understand which countries have the most users interested in Science.

It provides insights that can be useful for research, education, and targeted content creation related to scientific topics.

By sorting the results, we can identify the global distribution of science interest and focus on countries where science engagement is most prominent.


```
pd.read_sql_query('''SELECT Country, COUNT(*) AS Science_User_Count
FROM Social
WHERE Interests LIKE '%Science%'
GROUP BY Country
ORDER BY Science_User_Count DESC;''', conn)
```

This analysis identifies individuals born between 1950 and 1980 who have an interest in Fashion.

This analysis helps us identify individuals within a specific age group who are interested in Fashion.

By sorting alphabetically, it allows for easy comparison of Fashion interest across users in different countries.

It can be useful for targeting specific marketing campaigns or understanding fashion preferences across different age groups.

```
pd.read_sql_query('''SELECT Name, Country,  Interests
FROM Social
WHERE DOB BETWEEN '1950-01-01' AND '1980-12-31'
AND Interests LIKE '%Fashion%'
ORDER BY Name ASC;''', conn)
```

This analysis focuses on the interests of users from Canada.

This analysis identifies common interests between users from Canada and China.

his analysis helps us understand cultural similarities between Canada and China in terms of interests.

It can be valuable for international businesses, social research, and trend analysis.

By identifying shared interests, companies can develop strategies that appeal to both markets.


```
pd.read_sql_query('''SELECT Interests, COUNT(*) AS Total_Users
FROM (
    SELECT
        UserID,
       Interests,
        Country
    FROM social
) sub
WHERE Country IN ('Canada', 'China')
GROUP BY Interests
HAVING COUNT(DISTINCT Country) = 2 -- İlgi alanının her iki ülkede de olması gerekiyor
ORDER BY Total_Users DESC;''', conn)
```
